```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

def process_metrics_pipeline(df, metric_col='metric', metric_type_col='metric_type', file_id_col='file_id', method_col = 'integration_method', var_mask_col = 'integration_var_mask', dataset_col="dataset",
                             score_col='score', batch_col='batch', batch_correction_weight=0.5, bio_conservation_weight=0.5):
    """
    Generalized function to process a metrics pipeline DataFrame.

    Parameters:
        df (pd.DataFrame): Input DataFrame with required metric columns.
        metric_col (str): Column name representing the metric.
        metric_type_col (str): Column name representing the metric type.
        file_id_col (str): Column name representing the file ID.
        score_col (str): Column name representing the metric score.
        batch_col (str): Column name representing the batch.
        batch_correction_weight (float): Weight for batch correction in total score.
        bio_conservation_weight (float): Weight for bio conservation in total score.

    Returns:
        pd.DataFrame: Processed DataFrame with computed scores.
    """
    # Set index for transformation
    df = df[[metric_col, metric_type_col, file_id_col, method_col, score_col]].set_index(metric_col)
    
    # Apply MinMaxScaler per metric
    scaler = MinMaxScaler()
    #df['score_scaled'] = df.groupby(level=metric_col)[score_col]#transform(lambda x: scaler.fit_transform(x.values.reshape(-1, 1)).flatten())
    df['score_scaled'] = df[score_col]
    #print(df)
    
    # Reset index
    df.reset_index(inplace=True)

    # Create 'file_id_batch' column
    df[file_id_col] = df[file_id_col].astype(str)
    #df['file_id_batch'] = df[file_id_col].str.split('_').str[0] + '_batch_' + df[batch_col]
    df['file_id_batch'] = df[file_id_col]#df[method_col] + '_' + df[dataset_col] + '_' + df[var_mask_col]


    # Pivot table
    print(df)
    metrics_pivot = df.pivot_table(index='integration_method', columns=[metric_col], values='score_scaled')

    # Extract metric types
    temp_df = df[[metric_col, metric_type_col]].drop_duplicates()
    metric_type_dict = dict(zip(temp_df[metric_col], temp_df[metric_type_col]))

    # Identify batch_correction and bio_conservation metrics
    batch_correction_metrics = [metric for metric, category in metric_type_dict.items() if category == 'batch_correction']
    bio_conservation_metrics = [metric for metric, category in metric_type_dict.items() if category == 'bio_conservation']

    

    # Apply MinMaxScaler on batch correction and bio conservation metrics
    if batch_correction_metrics + bio_conservation_metrics:
        metrics_pivot.loc[:, batch_correction_metrics + bio_conservation_metrics] = scaler.fit_transform(
            metrics_pivot[batch_correction_metrics + bio_conservation_metrics]
        )

    # Morans' I inversion
    metrics_pivot['M\'s I:batch_covar_split'] = 1-metrics_pivot['M\'s I:batch_covar_split']

    # Compute aggregate scores
    metrics_pivot['Batch Correction'] = metrics_pivot[batch_correction_metrics].mean(axis=1)
    metrics_pivot['Bio Conservation'] = metrics_pivot[bio_conservation_metrics].mean(axis=1)

    # Compute weighted total score
    metrics_pivot['Total'] = (batch_correction_weight * metrics_pivot['Batch Correction'] +
                              bio_conservation_weight * metrics_pivot['Bio Conservation'])

    # Update metric type dictionary
    metric_type_dict.update({'Batch Correction': 'Aggregate Score', 'Bio Conservation': 'Aggregate Score', 'Total': 'Aggregate Score'})
    
    # Append metric_type row
    metrics_pivot.loc['metric_type'] = metrics_pivot.columns.map(metric_type_dict)

    # Clean index formatting
    metrics_pivot.index = metrics_pivot.index.str.replace('_manual_genes_extended', '', regex=False)

    return metrics_pivot
```

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import scanpy as sc
from anndata import AnnData
from plottable import ColumnDefinition, Table
from plottable.cmap import normed_cmap
from plottable.plots import bar
from sklearn.preprocessing import MinMaxScaler

def plot_results_table(
    df: pd.DataFrame,
    batch_correction_metrics: list[str] | None = None,
    bio_conservation_metrics: list[str] | None = None,
    min_max_scale: bool = False,
    show: bool = True,
    save_dir: str | None = None,
) -> Table:
    """
    Plot the benchmarking results.

    Parameters
    ----------
    df : pd.DataFrame
        The benchmarking results DataFrame.
    batch_correction_metrics : list[str] | None
        List of batch correction metrics.
    bio_conservation_metrics : list[str] | None
        List of bio conservation metrics.
    min_max_scale : bool, default=True
        Whether to apply MinMax scaling to the results.
    show : bool, default=True
        Whether to display the plot.
    save_dir : str | None, default=None
        Directory to save the plot. If `None`, the plot is not saved.

    Returns
    -------
    Table
        A table visualization of the benchmarking results.
    """
    
    num_embeds = len(df.index)
    _METRIC_TYPE = 'metric_type'
    # Apply MinMax scaling if requested
    if min_max_scale:
        df = pd.DataFrame(
            MinMaxScaler().fit_transform(df[:-1]),
            columns=df[:-1].columns,
            index=df[:-1].index,
        )
    #temp_df = pd.DataFrame(metrics_pipeline.groupby('metric')['metric_type'].first()).reset_index()
    #metric_type_dict.update({'Batch Correction': 'Aggregate Score', 'Bio Conservation': 'Aggregate Score', 'Total': 'Aggregate Score'})
    #df.loc['metric_type'] = df.columns.map(metric_type_dict)
    # Define the colormap function
    cmap_fn = lambda col_data: normed_cmap(col_data, cmap=plt.cm.PRGn, num_stds=2.5)

    # Drop the metric type row if it exists
    if _METRIC_TYPE in df.index:
        plot_df = df.drop(_METRIC_TYPE, axis=0)
    else:
        plot_df = df.copy()

    # Determine sorting column
    if batch_correction_metrics is not None and bio_conservation_metrics is not None:
        sort_col = "Total"
    elif batch_correction_metrics is not None:
        sort_col = "Batch Correction"
    else:
        #sort_col = "Bio Conservation"
        sort_col='Total'

    # Sort by the total score
    plot_df = plot_df.sort_values(by=sort_col, ascending=False).astype(np.float64)
    #print(plot_df)
    plot_df["Method"] = plot_df.index#.str.split('_').str[0]

    print(df)
    # Identify aggregate and other columns
    score_cols = df.columns[df.loc[_METRIC_TYPE] == 'Aggregate Score']
    print(score_cols)
    other_cols = df.columns[df.loc[_METRIC_TYPE] != 'Aggregate Score']

    # Column definitions for the table
    column_definitions = [
        ColumnDefinition("Method", width=1.5, textprops={"ha": "center", "weight": "bold"}),
    ]

    # Circles for metric values
    column_definitions += [
        ColumnDefinition(
            col,
            title=col.replace(" ", "\n", 1),
            width=1,
            textprops={
                "ha": "center",
                "bbox": {"boxstyle": "circle", "pad": 0.25},
            },
            cmap=cmap_fn(plot_df[col]),
            group=df.loc[_METRIC_TYPE, col],
            formatter="{:.2f}",
        )
        for col in other_cols
    ]

    # Bars for aggregate scores
    column_definitions += [
        ColumnDefinition(
            col,
            width=1,
            title=col.replace(" ", "\n", 1),
            plot_fn=bar,
            plot_kw={
                "cmap": plt.cm.YlGnBu,
                "plot_bg_bar": False,
                "annotate": True,
                "height": 0.9,
                "formatter": "{:.2f}",
            },
            group=df.loc[_METRIC_TYPE, col],
            border="left" if i == 0 else None,
        )
        for i, col in enumerate(score_cols)
    ]

    #print(column_definitions)

    # Create the table plot
    with plt.rc_context({"svg.fonttype": "none"}):
        fig, ax = plt.subplots(figsize=(len(df.columns) * 2.25, 3 + 0.5 * num_embeds))
        tab = Table(
            plot_df,
            cell_kw={"linewidth": 0, "edgecolor": "k"},
            column_definitions=column_definitions,
            ax=ax,
            row_dividers=True,
            footer_divider=True,
            textprops={"fontsize": 11, "ha": "center"},
            row_divider_kw={"linewidth": 1, "linestyle": (0, (1, 5))},
            col_label_divider_kw={"linewidth": 1, "linestyle": "-"},
            column_border_kw={"linewidth": 1, "linestyle": "-"},
            index_col="Method",
        ).autoset_fontcolors(colnames=plot_df.columns)

    # Show the plot
    if show:
        plt.show()

    # Save the figure if needed
    if save_dir is not None:
        fig.savefig("/lustre/groups/ml01/workspace/hpca/hpca_downstream/26_healthy_integration_metrics.svg", facecolor=ax.get_facecolor(), dpi=300)

    return tab
```
