```python
from adjustText import adjust_text
group="Alpha-Beta-Delta Cell"
genes_of_interest=["INS","GCG","SLC30A8","IAPP","PCSK1","PCSK2","CHGA","CHGB","PAX6","ISL1","SCG2","SCG3","SCGN"]
df=sc.get.rank_genes_groups_df(adata,group=group)
df=df.replace([np.inf,-np.inf],np.nan)
df=df.dropna(subset=["names","logfoldchanges","pvals_adj"])
df["pvals_adj_capped"]=df["pvals_adj"].clip(lower=1e-50)
df["neg_log10_padj"]=-np.log10(df["pvals_adj_capped"])
padj_cutoff=0.05
lfc_cutoff=1
df["category"]="Not significant"
df.loc[(df["pvals_adj"]<padj_cutoff)&(df["logfoldchanges"]>lfc_cutoff),"category"]="Up"
df.loc[(df["pvals_adj"]<padj_cutoff)&(df["logfoldchanges"]<-lfc_cutoff),"category"]="Down"
df["highlight"]=df["names"].isin(genes_of_interest)
colors={"Not significant":"lightgray","Up":"tab:red","Down":"tab:blue"}
plt.figure(figsize=(8,6))
for cat in["Not significant","Down","Up"]:
    sub=df[df["category"]==cat]
    plt.scatter(sub["logfoldchanges"],sub["neg_log10_padj"],s=8,alpha=0.45,c=colors[cat],label=cat,rasterized=True)
hi=df[df["highlight"]]
plt.scatter(hi["logfoldchanges"],hi["neg_log10_padj"],s=70,c="orange",edgecolor="black",linewidth=0.5,label="Genes of interest",zorder=5)
texts=[]
for _,row in hi.iterrows():
    texts.append(plt.text(row["logfoldchanges"],row["neg_log10_padj"],row["names"],fontsize=9,zorder=6))
adjust_text(texts,arrowprops=dict(arrowstyle="-",linewidth=0.5),expand_points=(1.2,1.4),expand_text=(1.2,1.4))
plt.axhline(-np.log10(padj_cutoff),linestyle="--",color="black",linewidth=1)
plt.axvline(-lfc_cutoff,linestyle="--",color="black",linewidth=1)
plt.axvline(lfc_cutoff,linestyle="--",color="black",linewidth=1)
plt.xlabel("log fold change")
plt.ylabel("-log10 adjusted p-value")
plt.title(f"Volcano plot: {group}")
plt.ylim(0,55)
plt.xlim(-12,12)
plt.legend(frameon=False)
plt.tight_layout()
plt.show()
```
