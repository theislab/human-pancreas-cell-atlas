# GSE167880 from Diego

```python
import os
import pandas as pd
import scanpy as sc
import anndata as ad

data_dir = "GSE167880"

features_file = os.path.join(data_dir, "GSE167880_features.tsv.gz")

features = pd.read_csv(features_file, sep="\t", header=None)

if features.shape[1] == 1:
    gene_symbols = features.iloc[:, 0].astype(str).values
    gene_ids = gene_symbols
else:
    gene_ids = features.iloc[:, 0].astype(str).values
    gene_symbols = features.iloc[:, 1].astype(str).values

sample_meta = {
    "GSM5114459": ("S5_DT", "Stage5"),
    "GSM5114460": ("S5_NT", "Stage5"),
    "GSM5114461": ("S6_A11", "Week0"),
    "GSM5114462": ("S6_B3", "Week0"),
    "GSM5114463": ("S6_B5", "Week0"),
    "GSM5114464": ("S7_D20", "Week3"),
    "GSM5114465": ("S7_D25", "Week3"),
    "GSM5114466": ("S7_D18", "Week3"),
    "GSM5114467": ("S7_exp22", "Week6"),
    "GSM5114468": ("S7_exp23", "Week6"),
    "GSM5114469": ("M1_A7", "Month1"),
    "GSM5114470": ("M1_H1", "Month1"),
    "GSM5114471": ("M1_H6", "Month1"),
    "GSM5114472": ("M3_B1", "Month3"),
    "GSM5114473": ("M3_E1", "Month3"),
    "GSM5114474": ("M3_E7", "Month3"),
    "GSM5114475": ("M6_CTRL", "Month6_CTRL"),
    "GSM5114476": ("M6_STZ", "Month6_STZ"),
}

adatas = []

for gsm, (sample_short, stage) in sample_meta.items():

    print("Reading", gsm)

    matrix_file = os.path.join(
        data_dir,
        f"{gsm}_{sample_short}_matrix.mtx.gz"
    )

    barcode_file = os.path.join(
        data_dir,
        f"{gsm}_{sample_short}_barcodes.tsv.gz"
    )

    # read matrix
    adata = sc.read_mtx(matrix_file).T

    # barcodes
    barcodes = pd.read_csv(
        barcode_file,
        sep="\t",
        header=None
    )[0].astype(str).values

    adata.obs_names = [f"{gsm}_{x}" for x in barcodes]

    # genes
    adata.var_names = gene_symbols
    adata.var["gene_ids"] = gene_ids
    adata.var_names_make_unique()

    # metadata
    adata.obs["GEO"] = gsm
    adata.obs["Sample_id"] = sample_short
    adata.obs["Stage"] = stage
    adata.obs["Dataset"] = "GSE167880"

    adatas.append(adata)

adata = ad.concat(
    adatas,
    join="outer",
    label="batch",
    index_unique=None
)

print(adata)

adata.write_h5ad(
    "GSE167880/GSE167880_sc_islets_raw.h5ad"
)
```

# QC

```python
adata = sc.read_h5ad('GSE167880/GSE167880_sc_islets_raw.h5ad')
```

```python
adata.X.max()
```

```python
sc.pp.calculate_qc_metrics(adata, inplace=True)

initial_cell_n = adata.n_obs

# Filter cells: Keep only cells with at least 1,000 UMIs and at least 200 genes expressed
sc.pp.filter_cells(adata, min_counts=1000)
sc.pp.filter_cells(adata, min_genes=200)

print(f"Filtered {initial_cell_n - adata.n_obs} cells; {adata.n_obs} remain after QC.")
```

```python
# Remove cells where mitochondrial (MT) gene counts are more than 25% of the total counts
if 'pct_counts_mt' not in adata.obs.columns:
    # For Scanpy/AnnData convention, mitochondrial genes usually start with 'MT-'
    mt_gene_mask = adata.var_names.str.upper().str.startswith('MT-')
    adata.obs['pct_counts_mt'] = (adata[:, mt_gene_mask].X.sum(axis=1).A1 if hasattr(adata[:, mt_gene_mask].X, "A1")
                                  else adata[:, mt_gene_mask].X.sum(axis=1)) / adata.X.sum(axis=1) * 100

mt_mask = adata.obs['pct_counts_mt'] <= 25
filtered_n = adata.n_obs - mt_mask.sum()

adata = adata[mt_mask].copy()
print(f"Filtered {filtered_n} cells with >25% mitochondrial reads; {adata.n_obs} remain after mtQC.")
```

```python
import scrublet as scr
import numpy as np

# For raw counts, use .X if raw is not present; otherwise adata.raw.X
counts_matrix = adata.X if not hasattr(adata, 'raw') or adata.raw is None else adata.raw.X

# Convert sparse matrix to dense if needed
if not isinstance(counts_matrix, np.ndarray):
    # Works for scipy.sparse matrices (e.g. csr_matrix, csc_matrix)
    counts_matrix = counts_matrix.toarray()

scrub = scr.Scrublet(counts_matrix)
doublet_scores, predicted_doublets = scrub.scrub_doublets()

# Optionally show Scrublet plot for QC
scrub.plot_histogram()

# Add doublet flag to AnnData object
adata.obs['doublet_score'] = doublet_scores
adata.obs['predicted_doublet'] = predicted_doublets

# Filter out predicted doublets
initial_n = adata.n_obs
adata = adata[~adata.obs['predicted_doublet']].copy()
print(f"Filtered {initial_n - adata.n_obs} predicted doublets using Scrublet; {adata.n_obs} cells remain.")
```

```python
# Define mapping for sample metadata: Sample_ID and Stage
sample_meta = {
    "GSM5114459": ("S5_DT", "Stage5"),
    "GSM5114460": ("S5_NT", "Stage5"),

    "GSM5114461": ("S7_A11", "Week0"),
    "GSM5114462": ("S7_B3", "Week0"),
    "GSM5114463": ("S7_B5", "Week0"),

    "GSM5114464": ("S7_D20", "Week3"),
    "GSM5114465": ("S7_D25", "Week3"),
    "GSM5114466": ("S7_D18", "Week3"),

    "GSM5114467": ("S7_exp22", "Week6"),
    "GSM5114468": ("S7_exp23", "Week6"),

    "GSM5114469": ("M1_A7", "Month1"),
    "GSM5114470": ("M1_H1", "Month1"),
    "GSM5114471": ("M1_H6", "Month1"),

    "GSM5114472": ("M3_B1", "Month3"),
    "GSM5114473": ("M3_E1", "Month3"),
    "GSM5114474": ("M3_E7", "Month3"),

    "GSM5114475": ("M6_CTRL", "Month6_CTRL"),
    "GSM5114476": ("M6_STZ", "Month6_STZ"),
}

# Update adata.obs with Sample_ID and Stage columns using the mapping
import pandas as pd

# Extract GSM ID from adata.obs_names (row/sample/cell names)
# Assumes GSM ID is the first part before "_" in each obs_name
gsm_ids = adata.obs_names.str.split("_").str[0]

# Map Sample_ID and Stage using the sample_meta dictionary
adata.obs['Sample_ID'] = gsm_ids.map(lambda gsm: sample_meta[gsm][0] if gsm in sample_meta else pd.NA)
adata.obs['Stage'] = gsm_ids.map(lambda gsm: sample_meta[gsm][1] if gsm in sample_meta else pd.NA)
```
