```python
import scanpy as sc
import pandas as pd
import numpy as np
```

```python
from glob import glob
```

```python
files = glob('/lustre/groups/ml01/workspace/hpca/data/*/*_concat.h5ad')
```

```python
files
```

```python
adatas = {}
for i in files:
    adatas[i] = sc.read_h5ad(i)
```

```python
for i in adatas.keys():
    if ('CellTypes' in adatas[i].obs) == False:
        adatas[i].obs['CellTypes'] = np.nan
```

```python
for i in adatas.keys():
    adatas[i].obs_names = adatas[i].obs_names + "_" + adatas[i].obs.ID.astype('str')
```

```python
for i in adatas.keys():
    adatas[i].var_names = adatas[i].var_names.astype('str')
    adatas[i].var_names_make_unique()
```

```python
for i in adatas.keys():
    del adatas[i].obsm
```

```python
for i in adatas.keys():
    print(i)
    print(adatas[i].X[1:50,1:50])
```

```python
for i in adatas.keys():
    print(i)
    print(adatas[i].raw)
    print(adatas[i].layers)
```

```python
for i in adatas.keys():
    import scipy.sparse as sp
    print(i)
    # Check if adata.X is dense (i.e., numpy array) and convert to sparse (CSR format)
    if not isinstance(adatas[i].X, sp._csr.csr_matrix):
        print(type(adatas[i].X))
        adatas[i].X = sp.csr_matrix(adatas[i].X)
```

```python
adata = sc.concat(adatas, )
```

```python
adata.write_h5ad('/lustre/groups/ml01/workspace/hpca/data/concatenated_all.h5ad')
```
