```python
adata_all.obs.Manual_Annotation.unique()
```

```python
adata_all.obs['Level_1_Harmonized'] = adata_all.obs.CellTypes.replace({
    'Endothelial': 'Endothelial Cells',
    'Acinar': 'Acinar Cell',
    'Ductal': 'Ductal Cell',
    'Endocrine': 'Endocrine Cell',
    'Lymphoid': 'Immune',  # Assuming B Cell for lymphoid
    'Myeloid': 'Macrophage',  # Myeloid lineages typically include macrophages
    'Stellate': 'Mesenchymal Cell',  # General stromal label for stellate cells
    #'Schwann': 'Mesenchymal Cell',  # Closest match given available labels
    'BloodVessel': 'Endothelial Cell',
    'LymphaticEndothelium': 'Endothelial Cell',
    'Fibroblast': 'Mesenchymal Cell',
    #'Immune': 'Macrophage/Endocrine Cell',  # General immune/stromal overlap
    'Adipocyte': 'Mesenchymal Cell',
    'Acinar-s': 'Acinar Cell',
    'Acinar-i': 'Acinar Cell',
    'Acinar-REG+': 'Acinar Cell',  # Noting contamination potential
    'Activated Stellate': 'Activated Stellate Cell',
    'Quiescent Stellate': 'Quiescent Stellate Cell',
    'Beta': 'Beta Cell',
    'Delta': 'Delta Cell',
    'MUC5B+ Ductal': 'Ductal Cell',  # Specialized subset of Ductal Cell
    'Macrophage': 'Macrophage',
    'Alpha': 'Alpha Cell',
    'Gamma': 'Pancreatic Polypeptide Cell'  # Known as gamma cells in pancreas

})
```

```python
adata_all.obs['celltype_all'] = adata_all.obs['Manual_Annotation'].copy()
```

```python
adata_all.obs.celltype_all = adata_all.obs.celltype_all.fillna(adata_all.obs.Level_1_Harmonized)
```

```python
karakose.obs['celltype_all'] = karakose.obs.Manual_Annotation
```

```python
del karakose.layers['scaledata']
```
